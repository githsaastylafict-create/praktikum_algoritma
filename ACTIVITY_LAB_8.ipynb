Aktivitas 3 - program Modular: Persegi Panjang

```python
def hitung_luas(panjang, lebar):
  return panjang * lebar

def hitung_keliling(panjang, lebar):
  return 2 * (panjang + lebar)

panjang = 8
lebar = 3
luas = hitung_luas(panjang, lebar)
keliling = hitung_keliling(panjang, lebar)

print("Luas:", luas)
print("keliling:", keliling)
```

karena ada perbedaan penggunaan rumus. untuk rumus luas adalah p*l, sedangjan rumus keliling adalah 2*(p+l). kalau di satukan nanti hasil nilainya error.

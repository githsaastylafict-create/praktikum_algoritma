```python

print("==============================")
print("       Githsa Flower          ")
print("==============================")

print("Pilihan Bunga:")
print("1. Teratai   - Rp20.000")
print("2. Tulip     - Rp17.000")
print("3. Matahari  - Rp21.000")

pilihan = input("Masukkan pilihan bunga: ")
jumlah = int(input("Masukkan jumlah beli: "))

if pilihan == "1":
    nama_bunga = "Teratai"
    harga = 20000
elif pilihan == "2":
    nama_bunga = "Tulip"
    harga = 17000
elif pilihan == "3":
    nama_bunga = "Matahari"
    harga = 21000
else:
    print("Pilihan bunga tidak tersedia!")
    exit()

kartu = input("Tambah kartu ucapan (Ya/Tidak): ")

if kartu.lower() == "ya":
    harga_kartu = 2000
else:
    harga_kartu = 0

total = (harga * jumlah) + harga_kartu

if total >= 100000:
    diskon = total * 10 / 100
else:
    diskon = 0

total_bayar = total - diskon

print("\n==============================")
print("===== STRUK PEMBELIAN BUNGA ==")
print("==============================")
print("Nama bunga  :", nama_bunga)
print("Jumlah      :", jumlah)
print("Harga Bunga : Rp", harga * jumlah)
print("Kartu       : Rp", harga_kartu)
print("Diskon      : Rp", int(diskon))
print("Total Bayar : Rp", int(total_bayar))
print("==============================")
print("======   TERIMA KASIH   ======")
print("==============================")
```
